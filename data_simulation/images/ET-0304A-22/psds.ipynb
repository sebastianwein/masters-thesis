```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import os
import sys
sys.path.append(os.path.abspath(os.path.join("/home/swein/Documents")))
```

```python
SMALL_SIZE = 8
MEDIUM_SIZE = 10
BIGGER_SIZE = 12

matplotlib.rc("font", size=MEDIUM_SIZE)          # controls default text sizes
matplotlib.rc("axes", titlesize=MEDIUM_SIZE)     # fontsize of the axes title
matplotlib.rc("axes", labelsize=MEDIUM_SIZE)    # fontsize of the x and y labels
matplotlib.rc("xtick", labelsize=SMALL_SIZE)    # fontsize of the tick labels
matplotlib.rc("ytick", labelsize=SMALL_SIZE)    # fontsize of the tick labels
matplotlib.rc("legend", fontsize=MEDIUM_SIZE)    # legend fontsize
matplotlib.rc("figure", titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
from pycbc.types.frequencyseries import FrequencySeries
from ggwd.utils.types import freq_series_from_numpy

low_freq_cutoff = 4
delta_f = 0.1
upper_freq = 5000
length = int((upper_freq - low_freq_cutoff)/delta_f)

ET10km_file_path = "18089_ET10kmcolumns.txt"
freq_data, hf_data, lf_data, lfhf_data = np.genfromtxt(ET10km_file_path, unpack=True)
ET10km_hf_psd = freq_series_from_numpy(freq_data, hf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
ET10km_lf_psd = freq_series_from_numpy(freq_data, lf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
ET10km_lfhf_psd = freq_series_from_numpy(freq_data, lfhf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)

ET15km_file_path = "18089_ET15kmcolumns.txt"
freq_data, hf_data, lf_data, lfhf_data = np.genfromtxt(ET15km_file_path, unpack=True)
ET15km_hf_psd = freq_series_from_numpy(freq_data, hf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
ET15km_lf_psd = freq_series_from_numpy(freq_data, lf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
ET15km_lfhf_psd = freq_series_from_numpy(freq_data, lfhf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)

ET20km_file_path = "18089_ET20kmcolumns.txt"
freq_data, hf_data, lf_data, lfhf_data = np.genfromtxt(ET20km_file_path, unpack=True)
ET20km_hf_psd = freq_series_from_numpy(freq_data, hf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
ET20km_lf_psd = freq_series_from_numpy(freq_data, lf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
ET20km_lfhf_psd = freq_series_from_numpy(freq_data, lfhf_data, length=length, delta_f=delta_f, low_freq_cutoff=low_freq_cutoff)
```

```python
fig, ax = plt.subplots(figsize=(13/2.54, 6/2.54))
ax.plot(ET10km_hf_psd.sample_frequencies, ET10km_hf_psd, ls="--", label="HF")
ax.plot(ET10km_lf_psd.sample_frequencies, ET10km_lf_psd, ls="--", label="LF")
ax.plot(ET10km_lfhf_psd.sample_frequencies, ET10km_lfhf_psd, label="LFHF")
ax.legend(frameon=False)
ax.set_xlim(low_freq_cutoff, upper_freq)
ax.set_xlabel("Frequency (Hz)")
ax.set_ylabel(r"Noise PSD ($\text{Hz}^{-1/2}$)")
ax.set_yscale("log")
ax.set_xscale("log")
fig.savefig("18089_ET10km.png", dpi=300, bbox_inches="tight")

fig, ax = plt.subplots(figsize=(13/2.54, 6/2.54))
ax.plot(ET10km_hf_psd.sample_frequencies, ET10km_hf_psd, ls="--", label="HF")
ax.plot(ET10km_lf_psd.sample_frequencies, ET10km_lf_psd, ls="--", label="LF")
ax.plot(ET10km_lfhf_psd.sample_frequencies, ET10km_lfhf_psd, label="LFHF")
ax.legend(frameon=False)
ax.set_xlim(low_freq_cutoff, upper_freq)
ax.set_xlabel("Frequency (Hz)")
ax.set_ylabel(r"Noise PSD ($\text{Hz}^{-1/2}$)")
ax.set_yscale("log")
ax.set_xscale("log")
fig.savefig("18089_ET15km.png")

fig, ax = plt.subplots(figsize=(13/2.54, 6/2.54))
ax.plot(ET10km_hf_psd.sample_frequencies, ET10km_hf_psd, ls="--", label="HF")
ax.plot(ET10km_lf_psd.sample_frequencies, ET10km_lf_psd, ls="--", label="LF")
ax.plot(ET10km_lfhf_psd.sample_frequencies, ET10km_lfhf_psd, label="LFHF")
ax.legend(frameon=False)
ax.set_xlim(low_freq_cutoff, upper_freq)
ax.set_xlabel("Frequency (Hz)")
ax.set_ylabel(r"Noise PSD ($\text{Hz}^{-1/2}$)")
ax.set_yscale("log")
ax.set_xscale("log")
fig.savefig("18089_ET20km.png")
```
