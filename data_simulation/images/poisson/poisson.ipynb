```python
import matplotlib
from matplotlib import pyplot as plt
import numpy as np
from scipy.special import factorial
```

```python
def poisson(x, lam, T):
    return (lam*T)**x/factorial(x)*np.exp(-lam*T) 
```

```python
SMALL_SIZE = 8
MEDIUM_SIZE = 10
BIGGER_SIZE = 12

matplotlib.rc("font", size=SMALL_SIZE)          # controls default text sizes
matplotlib.rc("axes", titlesize=MEDIUM_SIZE)     # fontsize of the axes title
matplotlib.rc("axes", labelsize=SMALL_SIZE)    # fontsize of the x and y labels
matplotlib.rc("xtick", labelsize=SMALL_SIZE)    # fontsize of the tick labels
matplotlib.rc("ytick", labelsize=SMALL_SIZE)    # fontsize of the tick labels
matplotlib.rc("legend", fontsize=SMALL_SIZE)    # legend fontsize
matplotlib.rc("figure", titlesize=MEDIUM_SIZE)  # fontsize of the figure title
```

```python
lam = 1E+6 / 3.154E+7
T = 64
x = np.arange(9)
p = poisson(x, lam, T)
```

```python
fig, ax = plt.subplots(figsize=(10/2.54, 6/2.54))
ax.bar(x, 100*p)
ax.set_xlabel("Number of signals")
ax.set_ylabel("Probability (%)")
# ax.set_yscale("log")
fig.tight_layout()
fig.savefig("poisson.png", dpi=300, bbox_inches="tight")
```
