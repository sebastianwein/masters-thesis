```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

```python
SMALL_SIZE = 8
MEDIUM_SIZE = 10
BIGGER_SIZE = 12

matplotlib.rc("font", size=SMALL_SIZE)          # controls default text sizes
matplotlib.rc("axes", titlesize=MEDIUM_SIZE)     # fontsize of the axes title
matplotlib.rc("axes", labelsize=SMALL_SIZE)    # fontsize of the x and y labels
matplotlib.rc("xtick", labelsize=SMALL_SIZE)    # fontsize of the tick labels
matplotlib.rc("ytick", labelsize=SMALL_SIZE)    # fontsize of the tick labels
matplotlib.rc("legend", fontsize=SMALL_SIZE)    # legend fontsize
matplotlib.rc("figure", titlesize=MEDIUM_SIZE)  # fontsize of the figure title
```

```python
def xy_plus(t, z, xy0, h_plus, w):
    x0, y0 = xy0
    x = x0 + x0/2*h_plus*np.sin(w*(t-z/(3E+8)))
    y = y0 - y0/2*h_plus*np.sin(w*(t-z/(3E+8)))
    return (x, y)

def xy_cross(t, z, xy0, h_cross, w):
    x0, y0 = xy0
    x = x0 + y0/2*h_cross*np.sin(w*(t-z/(3E+8)))
    y = y0 + x0/2*h_cross*np.sin(w*(t-z/(3E+8)))
    return (x, y)
```

```python
z = 0
h_plus, h_cross = 1, 1
w = 1
```

```python
n = 8
fig, ax = plt.subplots(ncols=n, figsize=(15/2.54, 2/2.54))
for i, t in enumerate(np.arange(n)*np.pi/(n/4)):
    theta = np.linspace(0, 1, 100)*2*np.pi
    xy0 = (np.cos(theta), np.sin(theta))
    x, y = xy0
    x, y = xy_plus(t, z, xy0, h_plus, w)
    ax[i].plot(x, y, c="k", lw=0.5, clip_on=False)
    theta = np.linspace(0, 1, 9)*2*np.pi
    xy0 = (np.cos(theta), np.sin(theta))
    x, y = xy0
    x, y = xy_plus(t, z, xy0, h_plus, w)
    ax[i].scatter(x, y, c=f"C{i}", clip_on=False, s=20, zorder=10)
    a = 1.5
    ax[i].set_xlim(-a, a)
    ax[i].set_ylim(-a, a)
    ax[i].set_aspect("equal")
    if i==0: label = r"$\omega t = 0$"
    elif i==1: label = r"$\pi/2$"
    elif i==2: label = r"$\pi$"
    else: 
        if i%2==0: label = fr"${i//2}\pi$"
        else: label = fr"${i}\pi/2$"
    ax[i].set_title(label)
    if i==0: 
        ax[i].spines["top"].set_visible(False)
        ax[i].spines["right"].set_visible(False)
        ax[i].set_xticks([])
        ax[i].set_yticks([])
        ax[i].set_xlabel(r"x")
        ax[i].set_ylabel(r"y", rotation=0, va="center")
    else: 
        ax[i].set_axis_off()
fig.savefig("plus_polarization.png", dpi=300, bbox_inches="tight")
```

```python
n = 8
mesh_number = 12
mesh_magnitude = 0.15
a = 1.5
h_plus = 0.7
h_cross = h_plus
fig, axs = plt.subplots(nrows=2, ncols=n, figsize=(16/2.54, 5/2.54))
ax = axs[0]
for i, t in enumerate(np.arange(n)*np.pi/(n/4)):
    theta = np.linspace(0, 1, 100)*2*np.pi
    xy0 = (np.cos(theta), np.sin(theta))
    x, y = xy0
    x, y = xy_plus(t, z, xy0, h_plus, w)
    ax[i].plot(x, y, c="k", lw=0.5, clip_on=False)
    theta = np.linspace(0, 1, 9)*2*np.pi
    xy0 = (np.cos(theta), np.sin(theta))
    x, y = xy0
    x, y = xy_plus(t, z, xy0, h_plus, w)
    ax[i].scatter(x, y, c=f"C{i}", clip_on=False, s=20, zorder=10)
    arr = np.linspace(-a, a, mesh_number)[1:-1]
    x0, y0 = np.meshgrid(arr, arr)
    xy0 = np.array([x0.flatten(), y0.flatten()])
    times = np.linspace(-1, 1, 100)*np.pi/2
    for xy in np.array(xy0).T:
        x, y = xy_plus(times, z, xy, mesh_magnitude, w)
        ax[i].plot(x, y, c="k", lw=0.5, clip_on=False, alpha=0.5)
    ax[i].set_xlim(-a, a)
    ax[i].set_ylim(-a, a)
    ax[i].set_aspect("equal")
    if i==0: label = r"$\omega t = 0$"
    elif i==1: label = r"$\pi/2$"
    elif i==2: label = r"$\pi$"
    else: 
        if i%2==0: label = fr"${i//2}\pi$"
        else: label = fr"${i}\pi/2$"
    ax[i].set_title(label)
    ax[i].set_axis_off()
ax = axs[1]
for i, t in enumerate(np.arange(n)*np.pi/(n/4)):
    theta = np.linspace(0, 1, 100)*2*np.pi
    xy0 = (np.cos(theta), np.sin(theta))
    x, y = xy_cross(t, z, xy0, h_plus, w)
    ax[i].plot(x, y, c="k", lw=0.5, clip_on=False)
    theta = np.linspace(0, 1, 9)*2*np.pi
    xy0 = (np.cos(theta), np.sin(theta))
    x, y = xy_cross(t, z, xy0, h_plus, w)
    ax[i].scatter(x, y, c=f"C{i}", clip_on=False, s=20, zorder=10)
    arr = np.linspace(-a, a, mesh_number)[1:-1]
    x0, y0 = np.meshgrid(arr, arr)
    xy0 = np.array([x0.flatten(), y0.flatten()])
    times = np.linspace(-1, 1, 100)*np.pi/2
    for xy in np.array(xy0).T:
        x, y = xy_cross(times, z, xy, mesh_magnitude, w)
        ax[i].plot(x, y, c="k", lw=0.5, clip_on=False, alpha=0.5)
    ax[i].set_xlim(-a, a)
    ax[i].set_ylim(-a, a)
    ax[i].set_aspect("equal")
    if i==0: 
        ax[i].spines["top"].set_visible(False)
        ax[i].spines["right"].set_visible(False)
        ax[i].set_xticks([])
        ax[i].set_yticks([])
        ax[i].set_xlabel(r"x")
        ax[i].set_ylabel(r"y", rotation=0, va="center")
    else: 
        ax[i].set_axis_off()
fig.subplots_adjust(hspace=0, wspace=0)
fig.savefig("polarization_trajectories.png", dpi=300, bbox_inches="tight")
```
