```python
def to_datetime(label):
    time = label[2:].split("_")
    utc = "-".join(["20"+time[0][:2], time[0][2:4], time[0][4:6]])
    if len(time) > 1: 
        utc += "T"+":".join([time[1][:2], time[1][2:4], time[1][4:6]])
    return np.datetime64(utc)

def to_float(np_datetime):
    delta = np_datetime.astype(datetime.timedelta)      
    return delta.total_seconds()
```
